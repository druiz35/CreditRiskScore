### Column implicit types

```python
for col in sorted(df.columns):
    print(f"{col}:")
    print(df[col].head())
    print(df[col].tail())
    unique_values = df[col].unique()
    print(unique_values[:int(len(unique_values)*0.95)])
    print("\n")
```

#### Conclusions

### Null and empty values

```python
df_nulls = df.isnull()
null_counts = pd.DataFrame({
    "col": df_nulls.columns,
    "null_count": df_nulls.sum(),
    "null_proportion": df_nulls.sum() / df.shape[0]
}).sort_values(by="null_count", ascending=False)
null_counts
```
